```python
import pandas as pd
import numpy as np 
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
df = pd.read_csv(r"C:\Users\nicob\One Drive Uniandes\OneDrive - Universidad de los Andes\Doctorado\proyecto\DatosTotales\gaia_parallax5.csv")
```

```python
scaler = MinMaxScaler()
```

```python
rango = [5,7,10]
```

```python
for i in range(len(rango)):
    df_new = df.copy()
    if rango[i] == 10:
        variables = ['pmra','pmdec','parallax']
        df_new = df_new[(df_new['parallax']>=rango[i])]
    else: 
        variables = ['pmra','pmdec','parallax','ra','dec']
        df_new = df_new[(df_new['parallax']>=rango[i])&(df_new['parallax']<rango[i+1])]
    
    norm_variables = [i+'_norm' for i in variables]
    df_new[norm_variables] = scaler.fit_transform(df_new[variables])
    print(len(df_new))
    df_new.to_csv(r"C:\Users\nicob\One Drive Uniandes\OneDrive - Universidad de los Andes"+f"\\Doctorado\proyecto\datos_shell\gaia_parallax5_{i}.csv")
```
